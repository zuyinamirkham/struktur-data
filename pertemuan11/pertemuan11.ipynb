Representasi Graf dengan Adjacency List

```python
# Membuat graf tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D':['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}
# Menampilkan graf
def print_graph(graf):
  for node in graf:
    print(f" {node} -> {graf [node]}")

print_graph(graph)
```

Traversal Graf menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque ([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end="")
      visited.add (vertex)
      queue.extend([neighbor for neighbor in graph [vertex] if neighbor not in visited])

#Contoh pemanggilan
bfs(graph, 'A')
```

Traversal Graf menggunakan DFS

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end="")
    visited.add(start)
    for neighbor in graph[start]:
      dfs (graph, neighbor, visited)
# Contoh pemanggilan
dfs (graph, 'A')
```

```python
from collections import deque

#DATA DARI TUGAS 1
graph_baru = {
    'A': ['B', 'C', 'E'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'E'],
    'D': ['B', 'E'],
    'E': ['A', 'B', 'C', 'D']
}

#IMPLEMENTASI BFS
def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

#IMPLEMENTASI DFS
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

print("--- Implementasi BFS dan DFS ---")

print("BFS: ", end=" ")
bfs(graph_baru, 'A')

print("\nDFS: ", end=" ")
dfs(graph_baru, 'A')
```

```python
def find_path(graph, start, end, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        visited.add(start)
        if start == end:
            return [start]
        for neighbor in graph[start]:
            jalur_lanjutan = find_path(graph, neighbor, end, visited)
            if jalur_lanjutan is not None:
                return [start] + jalur_lanjutan

    return None

graph_baru = {
    'A': ['B', 'C', 'E'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'E'],
    'D': ['B', 'E'],
    'E': ['A', 'B', 'C', 'D']
}
find_path(graph_baru, 'A', 'E')
```

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)
    return visited


def is_connected(graph):

    simpul_awal = list(graph.keys())[0]
    hasil_kunjungan = dfs(graph, simpul_awal)
    if len(hasil_kunjungan) == len(graph):
        return True
    else:
        return False

graph_baru = {
    'A': ['B', 'C', 'E'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'E'],
    'D': ['B', 'E'],
    'E': ['A', 'B', 'C', 'D']
}


is_connected(graph_baru)
```
